### Exercise

Write a function called `rhombus` that draws a rhombus with a given side length and a given interior angle. For example, here's a rhombus with side length `50` and an interior angle of `60` degrees.

````python
from jupyturtle import make_turtle

def rhombus(t, length, angle):
    second_angle = 180 - angle
    
    for _ in range(2):
        t.forward(length)
        t.left(angle)
        t.forward(length)
        t.left(second_angle)
````

You can use the following code to test your function.

````python
t=make_turtle()
rhombus(t,50, 60)
````

Now write a more general function called `parallelogram` that draws a quadrilateral with parallel sides. Then rewrite `rectangle` and `rhombus` to use `parallelogram`.

````python
def parallelogram(t, side1, side2, angle):
    second_angle = 180 - angle
    for _ in range(2):
        t.forward(side1)
        t.left(angle)
        t.forward(side2)
        t.left(second_angle)
````

````python
def rectangle(t, width, height):
    # A rectangle is a parallelogram with a 90-degree angle
    parallelogram(t, width, height, 90)
````

### Exercise

Write an appropriately general set of functions that can draw shapes like this.

![](https://github.com/AllenDowney/ThinkPython/raw/v3/jupyturtle_pie.png)

Hint: Write a function called `triangle` that draws one triangular segment, and then a function called `draw_pie` that uses `triangle`.

````python
# triangle 
import math
def triangle(t, r, angle):
    side = 2 * r * math.sin(math.radians(angle) / 2)
    
    base_angle = (180 - angle) / 2
    
    t.forward(r)
    t.left(180 - base_angle)
    t.forward(side)
    t.left(180 - base_angle)
    t.forward(r)

    
t1 = make_turtle()
triangle(t1, 50, 60) # testing triangle
````

````python
# Solution goes here
def draw_pie(t, n, r):
    angle = 360 / n 
    for _ in range(n):
        triangle(t, r, angle)
        t.left(angle)
````

You can use the following code to test your functions.

````python
turtle = make_turtle()

# jump(-80)
turtle.pen_up()
turtle.forward(-80)
turtle.pen_down()

size = 40

# draw_pie(5, size)
draw_pie(turtle, 5, size)

# jump(2*size)
turtle.pen_up()
turtle.forward(2 * size)
turtle.pen_down()

# draw_pie(6, size)
draw_pie(turtle, 6, size)

# jump(2*size)
turtle.pen_up()
turtle.forward(2 * size)
turtle.pen_down()

# draw_pie(7, size)
draw_pie(turtle, 7, size)
````
