Class vs Type: A class is a blueprint you write; a type is the category of data it belongs to. In Python, they are basically the same thing

Object vs Instance: An object is the actual thing living in the computer's memory. An instance is just a way of saying that object belongs to a specific class. For example, "Poppy" is an object, and he is an instance of the "Dog" class

Variable vs Attribute: A variable is a simple name for a value. An attribute is a variable that lives inside an object. You usually see them written with a dot, like dog.name

Pure vs Impure Functions: Pure functions take an input and return an output without changing anything else in your program. They are predictable and easy to fix. Impure functions might change variables outside themselves or print things to the screen. They are harder to track but necessary for things like saving files or showing info to a user

F-string Format Specifiers: These are codes used to change how text looks. For example, ":.2f" rounds a number to two decimal places, and ":," adds commas to big numbers to make them easier to read

```python
def subtract_time(t1, t2):
    """Compute the difference between two times in seconds.

    >>> subtract_time(make_time(3, 2, 1), make_time(3, 2, 0))
    1
    >>> subtract_time(make_time(3, 2, 1), make_time(3, 0, 0))
    121
    >>> subtract_time(make_time(11, 12, 0), make_time(9, 40, 0))
    5520
    """
    return None
```

```python
def subtract_time(t1, t2):
    # We make them into seconds
    # 3600 seconds in an hour, 60 seconds in a minute, seconds don't need multiplication
    seconds1 = t1.hour * 3600 + t1.minute * 60 + t1.second
    seconds2 = t2.hour * 3600 + t2.minute * 60 + t2.second
    return seconds1 - seconds2
```

Write a function called `is_after` that takes two `Time` objects and returns `True` if the first time is later in the day than the second, and `False` otherwise.

Here's an outline of the function to get you started.

```python
def is_after(t1, t2):
    """Checks whether `t1` is after `t2`.

    >>> is_after(make_time(3, 2, 1), make_time(3, 2, 0))
    True
    >>> is_after(make_time(3, 2, 1), make_time(3, 2, 1))
    False
    >>> is_after(make_time(11, 12, 0), make_time(9, 40, 0))
    True
    """
    return None
```

Here's a definition for a `Date` class that represents a date -- that is, a year, month, and day of the month.

```python
class Date:
    """Represents a year, month, and day"""
```

1. Write a function called `make_date` that takes `year`, `month`, and `day` as parameters, makes a `Date` object, assigns the parameters to attributes, and returns the result the new object. Create an object that represents June 22, 1933.

3. Write a function called `is_after` that takes two `Date` objects as parameters and returns `True` if the first comes after the second. Create a second object that represents September 17, 1933, and check whether it comes after the first object.

Hint: You might find it useful to write a function called `date_to_tuple` that takes a `Date` object and returns a tuple that contains its attributes in year, month, day order.

You can use this function outline to get started.

```python
def make_date(year, month, day):
    # Create date object
    date = Date() 
    
    # Assign attributes one by one
    date.year = year
    date.month = month
    date.day = day
    
    # Return the finished object
    return date
```

You can use this function outline to get started.

```python
def is_after(d1, d2):
    #check years first
    if d1.year > d2.year:
        return True
    if d1.year < d2.year:
        return False
    
    # If years are equal, check months
    if d1.month > d2.month:
        return True
    if d1.month < d2.month:
        return False
        
    # If months are also equal, check days
    return d1.day > d2.day
```
